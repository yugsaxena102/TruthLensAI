# 4.8 Text Exploration

## Objective

Before analyzing the vocabulary of fake and real news articles, a lightweight preprocessing step is performed.

The goal is not to prepare the text for machine learning, but rather to improve the readability of exploratory visualizations.

The preprocessing includes:

- Converting text to lowercase
- Removing punctuation
- Removing numbers
- Removing English stopwords
- Removing extra spaces

The original text remains unchanged.

```python
!pip -q install nltk
```

```python
# Section 4.8.1 – Create Clean Text for EDA

import re
import nltk

from nltk.corpus import stopwords

nltk.download("stopwords")

stop_words = set(stopwords.words("english"))



# Cleaning Function
def clean_text_for_eda(text):

    text = str(text).lower()

    # remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # remove punctuation
    text = re.sub(r"[^\w\s]", " ", text)

    # remove numbers
    text = re.sub(r"\d+", " ", text)

    # remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # remove stopwords
    words = [
        word
        for word in text.split()
        if word not in stop_words
    ]

    return " ".join(words)



# Apply
df["clean_text_eda"] = df["text"].apply(clean_text_for_eda)

print("EDA Text Created Successfully!")

display(
    df[
        ["text", "clean_text_eda"]
    ].head(3)
)
```

# 4.8.2 Most Frequent Words

## Objective

This section identifies the most frequently occurring words in fake and real news articles.

Unlike word clouds, frequency plots provide an exact quantitative comparison between the vocabularies of both classes.

The analysis helps identify dominant themes and recurring terminology used in each news category.

```python


from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def get_top_words(corpus, n=20):

    vectorizer = CountVectorizer()

    X = vectorizer.fit_transform(corpus)

    word_counts = X.sum(axis=0).A1

    words = vectorizer.get_feature_names_out()

    freq = pd.DataFrame({
        "word": words,
        "count": word_counts
    })

    return freq.sort_values(
        by="count",
        ascending=False
    ).head(n)



    # Generate Top Words

fake_top_words = get_top_words(
    df[df["label"] == 1]["clean_text_eda"]
)

real_top_words = get_top_words(
    df[df["label"] == 0]["clean_text_eda"]
)

display(fake_top_words)

display(real_top_words)


# Plot

fig, axes = plt.subplots(1, 2, figsize=(18,8))

sns.barplot(
    data=fake_top_words,
    y="word",
    x="count",
    ax=axes[0]
)

axes[0].set_title("Top 20 Words - Fake News")

sns.barplot(
    data=real_top_words,
    y="word",
    x="count",
    ax=axes[1]
)

axes[1].set_title("Top 20 Words - Real News")

plt.tight_layout()
plt.show()
```

### Key Insights

- Both fake and real news articles are heavily centered around U.S. political events, particularly the 2015–2017 presidential election.

- Real news articles contain institutional vocabulary such as *Reuters*, *government*, and *house*, indicating reliance on established news agencies.

- Fake news articles more frequently reference political personalities such as *Obama*, *Hillary*, and *Donald*, suggesting a stronger focus on individuals.

- Several extremely common words (e.g., *said*, *would*, *one*, *people*) dominate both classes. These words are unlikely to be discriminative and will be considered as custom stopwords for visualization purposes.

- The similarity of the top vocabulary across both classes highlights the need for contextual language models rather than relying solely on word frequency.

# 5.1 Data Cleaning & Preparation

In this section, the dataset is cleaned and prepared before model training.

The following preprocessing steps are performed:

- Remove unnecessary columns
- Handle missing values
- Combine the title and article text into a single feature
- Remove duplicate articles
- Verify the final dataset

```python
# ============================================================
# Data Cleaning & Preparation
# ============================================================

# Create a copy to preserve the original dataset
df_clean = df.copy()

# Drop unnecessary index column
df_clean.drop(columns=["Unnamed: 0"], inplace=True)

# Fill missing values
df_clean["title"] = df_clean["title"].fillna("")
df_clean["text"] = df_clean["text"].fillna("")

# Merge title and text
df_clean["combined_text"] = (
    df_clean["title"].str.strip()
    + " "
    + df_clean["text"].str.strip()
)

# Remove extra spaces
df_clean["combined_text"] = (
    df_clean["combined_text"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Remove duplicate articles
before = len(df_clean)

df_clean.drop_duplicates(
    subset=["combined_text"],
    inplace=True
)

after = len(df_clean)

print("="*70)
print("Data Cleaning Summary")
print("="*70)

print(f"Rows Before Cleaning : {before}")
print(f"Rows After Cleaning  : {after}")
print(f"Duplicates Removed   : {before-after}")

print("\nFinal Shape")
print(df_clean.shape)

print("\nColumns")
print(df_clean.columns.tolist())

print("\nSample Combined Text")
# Keep only the required columns
df_clean = df_clean[
    [
        "combined_text",
        "label"
    ]
]

print(df_clean.shape)
display(df_clean.head())
```

# 5.2 NLP Text Cleaning for Machine Learning

Machine Learning models cannot understand raw text directly. Therefore, the news articles are cleaned and normalized before feature extraction.

The preprocessing pipeline includes:

- Convert text to lowercase
- Remove HTML tags
- Remove URLs
- Remove Twitter mentions (@username)
- Remove hashtags (#) while keeping the word
- Remove special characters and punctuation
- Remove numbers
- Remove extra whitespaces
- Remove English stopwords
- Perform lemmatization

The cleaned text will later be converted into numerical vectors using TF-IDF.

```python
# ============================================================
# Import Libraries
# ============================================================

import re
import nltk

from bs4 import BeautifulSoup
from nltk.corpus import stopwords, wordnet
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

# Download required resources
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")   # for newer NLTK versions

# Initialize objects
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()


# ============================================================
# Text Cleaning Function
# ============================================================



# Convert NLTK POS tags to WordNet POS tags

def get_wordnet_pos(tag):

    if tag.startswith("J"):
        return wordnet.ADJ

    elif tag.startswith("V"):
        return wordnet.VERB

    elif tag.startswith("N"):
        return wordnet.NOUN

    elif tag.startswith("R"):
        return wordnet.ADV

    else:
        return wordnet.NOUN





def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove HTML tags
    text = BeautifulSoup(text, "html.parser").get_text()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", " ", text)

    # Remove Twitter mentions
    text = re.sub(r"@\w+", " ", text)

    # Remove hashtag symbol but keep the word
    text = re.sub(r"#", "", text)

    # Remove [VIDEO], [PHOTO], etc.
    text = re.sub(r"\[.*?\]", " ", text)

    # Remove punctuation and special characters
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenization
    words = text.split()

    # Remove stopwords
    words = [word for word in words if word not in stop_words]

    # POS tagging
    tagged_words = pos_tag(words)

    # POS-aware Lemmatization
    words = [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tagged_words
    ]

    return " ".join(words)
```

### Apply Text Cleaning

The cleaning function is applied to the combined news articles to create a new feature named `clean_text_ml`, which will be used for Machine Learning models.

```python
# ============================================================
# Apply Cleaning Function
# ============================================================

print("Cleaning text... This may take a few minutes.")

df_clean["clean_text_ml"] = df_clean["combined_text"].apply(clean_text)

print("\nCleaning Completed Successfully!")

display(
    df_clean[
        ["combined_text", "clean_text_ml"]
    ].head(3)
)
```

# 6.2 Model Evaluation Framework

Instead of evaluating each machine learning model separately, a reusable evaluation framework is created.

For every model, the framework automatically computes:

- Accuracy
- Precision
- Recall
- F1-Score
- ROC-AUC Score
- Confusion Matrix
- Classification Report
- Training Time
- Prediction Time

The evaluation results are stored for later comparison.

```python
# ============================================================
# Model Evaluation Framework
# ============================================================

import time
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# Store results of every model
results = []

def evaluate_model(model, model_name):

    print("=" * 70)
    print(f"Training {model_name}")
    print("=" * 70)

    # -----------------------------
    # Training
    # -----------------------------
    start_train = time.time()

    model.fit(X_train_tfidf, y_train)

    train_time = time.time() - start_train

    # -----------------------------
    # Prediction
    # -----------------------------
    start_pred = time.time()

    y_pred = model.predict(X_test_tfidf)

    pred_time = time.time() - start_pred

    # -----------------------------
    # Probability / Decision Scores
    # -----------------------------
    # Get prediction scores for ROC-AUC

    if hasattr(model, "predict_proba"):

        # Models like Logistic Regression, Naive Bayes,
        # Random Forest return class probabilities.

        y_score = model.predict_proba(X_test_tfidf)[:, 1]

    elif hasattr(model, "decision_function"):

        # Models like Linear SVM return decision scores.

        y_score = model.decision_function(X_test_tfidf)

    else:

        # Fallback for models without probability support.

        y_score = y_pred

    # -----------------------------
    # Metrics
    # -----------------------------
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_score)

    # Save results
    results.append({

        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC AUC": roc_auc,
        "Train Time (s)": train_time,
        "Prediction Time (s)": pred_time

    })

    # -----------------------------
    # Print Metrics
    # -----------------------------
    print(f"Accuracy        : {accuracy:.4f}")
    print(f"Precision       : {precision:.4f}")
    print(f"Recall          : {recall:.4f}")
    print(f"F1 Score        : {f1:.4f}")
    print(f"ROC-AUC         : {roc_auc:.4f}")

    print(f"\nTraining Time   : {train_time:.3f} sec")
    print(f"Prediction Time : {pred_time:.3f} sec")

    # -----------------------------
    # Classification Report
    # -----------------------------
    print("\nClassification Report\n")

    print(classification_report(y_test, y_pred))

    # -----------------------------
    # Confusion Matrix
    # -----------------------------
    cm = confusion_matrix(y_test, y_pred)

    plt.figure(figsize=(5,4))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Real","Fake"],
        yticklabels=["Real","Fake"]
    )

    plt.title(f"{model_name} Confusion Matrix")

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.show()




    results_df = pd.DataFrame(results)

    display(results_df)
```

# 7.3 Universal Hyperparameter Tuning Function

A reusable training function is created to automate the hyperparameter optimization process.

The function performs the following tasks:

- Selects Grid Search or Randomized Search automatically.
- Uses 5-Fold Stratified Cross Validation.
- Finds the best hyperparameters.
- Measures training time.
- Returns the optimized model for evaluation.

This modular design makes it easy to add additional machine learning models in the future.

```python
# ============================================================
# Universal Hyperparameter Tuning Function
# ============================================================

import time
import numpy as np

from sklearn.model_selection import (
    GridSearchCV,
    RandomizedSearchCV
)


def tune_model(model_name, config):

    print("=" * 70)
    print(f"Tuning : {model_name}")
    print("=" * 70)

    model = config["model"]
    params = config["params"]
    search_type = config["search_type"]

    # ============================================================
    # Multiple Evaluation Metrics
    # ============================================================

    scoring = {

        "Accuracy": "accuracy",
        "Precision": "precision",
        "Recall": "recall",
        "F1": "f1",
        "ROC_AUC": "roc_auc"

    }

    # ============================================================
    # Select Search Method
    # ============================================================

    if search_type == "grid":

        search = GridSearchCV(

            estimator=model,

            param_grid=params,

            scoring=scoring,

            refit="F1",

            cv=cv,

            n_jobs=-1,

            pre_dispatch="2*n_jobs",

            verbose=2,

            return_train_score=True,

            error_score="raise"

        )

    elif search_type == "random":

        search = RandomizedSearchCV(

            estimator=model,

            param_distributions=params,

            n_iter=10,

            scoring=scoring,

            refit="F1",

            cv=cv,

            random_state=42,

            n_jobs=1,

            pre_dispatch="2*n_jobs",

            verbose=2,

            return_train_score=True,

            error_score="raise"

        )

    else:

        raise ValueError(f"Unknown search type : {search_type}")

    # ============================================================
    # Model Training
    # ============================================================

    start = time.time()

    search.fit(X_train_tfidf, y_train)

    end = time.time()

    training_time = end - start

    # ============================================================
    # Best Result
    # ============================================================

    best_index = search.best_index_

    cv_results = search.cv_results_

    print("\nTraining Completed Successfully!")

    print(f"\nBest Cross Validation F1 : {search.best_score_:.4f}")

    print("\nBest Hyperparameters")
    print("-" * 40)

    for key, value in search.best_params_.items():
        print(f"{key:<20}: {value}")

    # ============================================================
    # Cross Validation Performance
    # ============================================================

    print("\nCross Validation Performance")
    print("-" * 40)

    print(f"Accuracy     : {cv_results['mean_test_Accuracy'][best_index]:.4f}")
    print(f"Precision    : {cv_results['mean_test_Precision'][best_index]:.4f}")
    print(f"Recall       : {cv_results['mean_test_Recall'][best_index]:.4f}")
    print(f"F1 Score     : {cv_results['mean_test_F1'][best_index]:.4f}")
    print(f"ROC-AUC      : {cv_results['mean_test_ROC_AUC'][best_index]:.4f}")

    # ============================================================
    # Training Statistics
    # ============================================================

    print("\nTraining Statistics")
    print("-" * 40)

    print(f"Training Time : {training_time:.2f} sec")
    print(f"Best Rank     : {cv_results['rank_test_F1'][best_index]}")
    print(f"Mean Train F1 : {cv_results['mean_train_F1'][best_index]:.4f}")
    print(f"Mean Valid F1 : {cv_results['mean_test_F1'][best_index]:.4f}")
    print(f"Std Dev F1    : {cv_results['std_test_F1'][best_index]:.4f}")

    print("\n" + "=" * 70)
    print(f"{model_name} Hyperparameter Tuning Completed Successfully")
    print("=" * 70)

    # ============================================================
    # Return Results
    # ============================================================

    return {

        "model_name": model_name,

        "best_model": search.best_estimator_,

        "best_params": search.best_params_,

        "search_object": search,

        "cv_accuracy": cv_results["mean_test_Accuracy"][best_index],

        "cv_precision": cv_results["mean_test_Precision"][best_index],

        "cv_recall": cv_results["mean_test_Recall"][best_index],

        "cv_f1": cv_results["mean_test_F1"][best_index],

        "cv_roc_auc": cv_results["mean_test_ROC_AUC"][best_index],

        "train_f1": cv_results["mean_train_F1"][best_index],

        "valid_f1": cv_results["mean_test_F1"][best_index],

        "std_f1": cv_results["std_test_F1"][best_index],

        "best_rank": cv_results["rank_test_F1"][best_index],

        "training_time": training_time

    }
```

- Best hyperparameters are returned.
- The optimized model is ready for evaluation.

```python
# ============================================================
# Universal Model Evaluation Function
# ============================================================

import time

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)


def evaluate_model(model_name, model, X_test, y_test):

    print("\n" + "=" * 70)
    print(f"Evaluating : {model_name}")
    print("=" * 70)

    # ============================================================
    # Prediction
    # ============================================================

    start = time.time()

    y_pred = model.predict(X_test)

    prediction_time = time.time() - start

    # ============================================================
    # Prediction Scores
    # ============================================================

    if hasattr(model, "predict_proba"):

        y_score = model.predict_proba(X_test)[:, 1]

    elif hasattr(model, "decision_function"):

        y_score = model.decision_function(X_test)

    else:

        y_score = y_pred

    # ============================================================
    # Metrics
    # ============================================================

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(y_test, y_pred)

    recall = recall_score(y_test, y_pred)

    f1 = f1_score(y_test, y_pred)

    roc_auc = roc_auc_score(y_test, y_score)

    # ============================================================
    # Print Results
    # ============================================================

    print("\nTest Performance")
    print("-" * 40)

    print(f"Accuracy        : {accuracy:.4f}")
    print(f"Precision       : {precision:.4f}")
    print(f"Recall          : {recall:.4f}")
    print(f"F1 Score        : {f1:.4f}")
    print(f"ROC-AUC         : {roc_auc:.4f}")
    print(f"Prediction Time : {prediction_time:.4f} sec")

    print("\nClassification Report")
    print("-" * 40)

    print(classification_report(y_test, y_pred))

    print("\nConfusion Matrix")
    print("-" * 40)

    cm = confusion_matrix(y_test, y_pred)

    print(cm)

    # ============================================================
    # Return Results
    # ============================================================

    return {

        "accuracy": accuracy,

        "precision": precision,

        "recall": recall,

        "f1": f1,

        "roc_auc": roc_auc,

        "prediction_time": prediction_time,

        "classification_report": classification_report(
            y_test,
            y_pred,
            output_dict=True
        ),

        "confusion_matrix": cm

    }
```

📌 7.5 Complete ML Training Pipeline

```python
# ============================================================
# Complete Machine Learning Pipeline
# ============================================================

results = []

best_models = {}

print("=" * 80)
print("Starting Machine Learning Pipeline")
print("=" * 80)

for model_name, config in model_configs.items():

    print("\n" + "=" * 80)
    print(f"Processing : {model_name}")
    print("=" * 80)

    # --------------------------------------------------------
    # Hyperparameter Tuning
    # --------------------------------------------------------

    tuning_result = tune_model(
        model_name=model_name,
        config=config
    )

    # --------------------------------------------------------
    # Model Evaluation
    # --------------------------------------------------------

    evaluation_result = evaluate_model(
        model_name=model_name,
        model=tuning_result["best_model"],
        X_test=X_test_tfidf,
        y_test=y_test
    )

    # --------------------------------------------------------
    # Save Best Model
    # --------------------------------------------------------

    best_models[model_name] = tuning_result["best_model"]

    # --------------------------------------------------------
    # Store Results
    # --------------------------------------------------------

    results.append({

        "Model": model_name,

        "CV Accuracy": tuning_result["cv_accuracy"],
        "CV Precision": tuning_result["cv_precision"],
        "CV Recall": tuning_result["cv_recall"],
        "CV F1": tuning_result["cv_f1"],
        "CV ROC-AUC": tuning_result["cv_roc_auc"],

        "Test Accuracy": evaluation_result["accuracy"],
        "Test Precision": evaluation_result["precision"],
        "Test Recall": evaluation_result["recall"],
        "Test F1": evaluation_result["f1"],
        "Test ROC-AUC": evaluation_result["roc_auc"],

        "Training Time (sec)": tuning_result["training_time"],
        "Prediction Time (sec)": evaluation_result["prediction_time"]

    })

print("\n" + "=" * 80)
print("All Models Processed Successfully!")
print("=" * 80)
```
